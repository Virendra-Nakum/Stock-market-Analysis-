```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

print("1. Load CSV File")

try:
    path = input("Enter Dataset Path (e.g. data/sales_data.csv): ").strip()
    data = pd.read_csv(path)
    print("Dataset Loaded Successfully!")
 
except :
    print("Invalid File Name:")
```

```python
print("\n========== EXPLORE DATASET ==========")
print("1. Display First 5 Rows")
print("2. Display Last 5 Rows")
print("3. Display Column Names")
print("4. Display Data Types")
print("5. Display Dataset Information")
print("6. Display Statistical Summary")

try:
    choice = int(input("Enter Your Choice: "))

    if choice == 1  :
        print(data.head())

    elif choice == 2 :
        print(data.tail())

    elif choice == 3 :
        print(data.columns)

    elif choice == 4 :
        print(data.dtypes)

    elif choice == 5 :
        print(data.info())

    elif choice == 6 :
        print(data.describe())

    else :
        print("Enter A Valid Choice !")

except :
    print("Something Went Wrong !")
```

```python
print("\n========== CLEAN DATASET ==========")
print("1. Check Missing Values")
print("2. Handle Missing Values")
print("3. Remove Duplicate Rows")
print("4. Convert Data Types")
print("5. Check Invalid Values")
try:
    choice = input("Enter Your Choice: ")

    if choice == "1":
        print(data.isnull().sum())

    elif choice == "2":
        data = data.ffill()
        print("Missing Values Filled!")

    elif choice == "3":
        data.drop_duplicates(inplace=True)
        print("Duplicate Rows Removed!")

    elif choice == "4":
        data["Date"] = pd.to_datetime(data["Date"])
        print(data.dtypes)

    elif choice == "5":
        print(data[data["Close"] <= 0])

    else:
        print("Invalid Choice!")

except :
    print("Somethig Went Wrong !")
```

```python
print("\n========== DATA ANALYSIS ==========")
print("1. Calculate Daily Returns")
print("2. Calculate Average Stock Price")
print("3. Find Highest and Lowest Price")
print("4. Calculate Stock Volatility")
print("5. Analyze Trading Volume")
print("6. Calculate Moving Average")
print("7. Analyze Daily Price Changes")

choice = input("Enter Your Choice: ")

if choice == "1":
    data["Daily_Return"] = data["Close"].pct_change() * 100
    print(data[["Date", "Close", "Daily_Return"]])

elif choice == "2":
    print("Average Open Price:", data["Open"].mean())
    print("Average High Price:", data["High"].mean())
    print("Average Low Price:", data["Low"].mean())
    print("Average Close Price:", data["Close"].mean())

elif choice == "3":
    print("Highest Price:", data["High"].max())
    print("Lowest Price:", data["Low"].min())

elif choice == "4":
    volatility = data["Close"].pct_change().std() * 100
    print("Stock Volatility:", volatility)

elif choice == "5":
    print("Total Volume:", data["Volume"].sum())
    print("Average Volume:", data["Volume"].mean())
    print("Highest Volume:", data["Volume"].max())

elif choice == "6":
    data["Moving_Average"] = data["Close"].rolling(7).mean()
    print(data[["Date", "Close", "Moving_Average"]])

elif choice == "7":
    data["Price_Change"] = data["Close"].diff()
    print(data[["Date", "Close", "Price_Change"]])

else:
    print("Invalid Choice!")
```

```python
print("\n========== DATA VISUALIZATION ==========")
print("1. Stock Price Line Chart")
print("2. Trading Volume Bar Chart")
print("3. Moving Average Chart")
print("4. Daily Returns Histogram")
print("5. Stock Price Area Chart")
print("6. Correlation Heatmap")

choice = input("Enter Your Choice: ").strip()

if data.empty:
    print("No Data Available!")

elif choice == "1" or choice == "2" or choice == "3" or choice == "4" or choice == "5" or choice == "6":

    if choice == "1" or choice == "3" or choice == "5":
        columns = ["Date", "Close"]
    elif choice == "2":
        columns = ["Date", "Volume"]
    elif choice == "4":
        columns = ["Close"]
    else:
        columns = ["Open", "High", "Low", "Close", "Volume"]

    if any(col not in data.columns for col in columns):
        print("Required Columns Not Found!")

    else:
        if "Date" in data.columns:
            data["Date"] = pd.to_datetime(data["Date"], errors="coerce")
            data = data.dropna(subset=["Date"])

        if choice == "1":
            sns.lineplot(data=data, x="Date", y="Close")
            plt.title("Stock Price Line Chart")

        elif choice == "2":
            sns.barplot(data=data, x="Date", y="Volume")
            plt.title("Trading Volume Bar Chart")

        elif choice == "3":
            data["Moving_Average"] = data["Close"].rolling(7).mean()
            sns.lineplot(data=data, x="Date", y="Close", label="Close Price")
            sns.lineplot(data=data, x="Date", y="Moving_Average", label="Moving Average")
            plt.title("Stock Moving Average")

        elif choice == "4":
            data["Daily_Return"] = data["Close"].pct_change() * 100
            sns.histplot(data=data, x="Daily_Return", bins=20, kde=True)
            plt.title("Daily Returns Histogram")

        elif choice == "5":
            plt.plot(data["Date"], data["Close"])
            plt.fill_between(data["Date"], data["Close"], alpha=0.3)
            plt.title("Stock Price Area Chart")

        elif choice == "6":
            sns.heatmap(data[columns].corr(), annot=True, cmap="coolwarm")
            plt.title("Stock Market Correlation")

        if choice != "4" and choice != "6":
            plt.xticks(rotation=45)
            plt.locator_params(axis="x", nbins=10)

        plt.tight_layout()
        plt.show()

else:
    print("Invalid Choice! Enter 1 to 6.")
```

```python
print("\n========== ANALYSIS RESULTS ==========")
print("1. Display Overall Summary")
print("2. Display Stock Performance")
print("3. Display Risk Analysis")
print("4. Display Trend Analysis")
print("5. Display Final Report")

choice = input("Enter Your Choice: ")

if choice == "1":
    print("\n========== OVERALL SUMMARY ==========")
    print("Total Trading Days:", len(data))
    print("Average Close Price:", data["Close"].mean())
    print("Highest Price:", data["High"].max())
    print("Lowest Price:", data["Low"].min())
    print("Total Trading Volume:", data["Volume"].sum())

elif choice == "2":
    print("\n========== STOCK PERFORMANCE ==========")
    first_price = data["Close"].iloc[0]
    last_price = data["Close"].iloc[-1]

    change = last_price - first_price
    percentage = (change / first_price) * 100

    print("Starting Price:", first_price)
    print("Latest Price:", last_price)
    print("Price Change:", change)
    print("Performance:", percentage, "%")

elif choice == "3":
    print("\n========== RISK ANALYSIS ==========")
    daily_returns = data["Close"].pct_change()
    volatility = daily_returns.std() * 100

    print("Stock Volatility:", volatility, "%")
    print("Maximum Daily Loss:", daily_returns.min() * 100, "%")
    print("Maximum Daily Gain:", daily_returns.max() * 100, "%")

elif choice == "4":
    print("\n========== TREND ANALYSIS ==========")
    data["Moving_Average"] = data["Close"].rolling(7).mean()

    if data["Close"].iloc[-1] > data["Moving_Average"].iloc[-1]:
        print("Stock Trend: Upward")
    else:
        print("Stock Trend: Downward")

    sns.lineplot(data=data, x="Date", y="Close", label="Close Price")
    sns.lineplot(data=data, x="Date", y="Moving_Average", label="Moving Average")
    plt.title("Stock Trend Analysis")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

elif choice == "5":
    print("\n========== FINAL STOCK REPORT ==========")
    print(data[["Open", "High", "Low", "Close", "Volume"]].describe())
    print("\nOverall Performance:")

    if data["Close"].iloc[-1] > data["Close"].iloc[0]:
        print("Stock Price Increased During This Period.")
    elif data["Close"].iloc[-1] < data["Close"].iloc[0]:
        print("Stock Price Decreased During This Period.")
    else:
        print("Stock Price Remained Unchanged.")

else:
    print("Invalid Choice!")
```
